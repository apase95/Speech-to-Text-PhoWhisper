```python
import torch
import transformers
from transformers import AutoProcessor

MODEL_NAME = "vinai/PhoWhisper-base"
print("torch", torch.__version__)
print("transformers", transformers.__version__)
print("cuda", torch.cuda.is_available())
processor = AutoProcessor.from_pretrained(MODEL_NAME)
print("model config ok", MODEL_NAME, len(processor.tokenizer))
```

```python
from transformers import pipeline

audio_path = "../data/raw/test/sample.wav"  # đổi sang file wav/mp3 của bạn
device = 0 if torch.cuda.is_available() else -1
dtype = torch.float16 if torch.cuda.is_available() else torch.float32
asr = pipeline("automatic-speech-recognition", model=MODEL_NAME, device=device, torch_dtype=dtype)
asr(audio_path, generate_kwargs={"language": "vi", "task": "transcribe"})["text"]
```
